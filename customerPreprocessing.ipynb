```python
# Function to remove outliers for all numeric columns
def remove_outliers(df):
    # Iterate through each numeric column
    for col in df.select_dtypes(include=['float64', 'int64']):
        Q1 = df[col].quantile(0.25)  # First quartile
        Q3 = df[col].quantile(0.75)  # Third quartile
        IQR = Q3 - Q1  # Interquartile range
        
        # Define outlier bounds
        lower_limit = Q1 - 1.5 * IQR
        upper_limit = Q3 + 1.5 * IQR
        
        # Filter out rows where the column value is outside the outlier bounds
        df = df[(df[col] >= lower_limit) & (df[col] <= upper_limit)]
        
    return df

# Remove outliers from the entire dataset
df = remove_outliers(df)

# Display the cleaned DataFrame
print(df)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# Convert ALL numeric columns to proper numeric types
df["repeat_purchase_rate"] = pd.to_numeric(df["repeat_purchase_rate"], errors="coerce")
df["customer_lifetime_value"] = pd.to_numeric(df["customer_lifetime_value"], errors="coerce")
df["avg_order_value"] = pd.to_numeric(df["avg_order_value"], errors="coerce")
df["purchase_frequency"] = pd.to_numeric(df["purchase_frequency"], errors="coerce")

# --------------------------------------
# 1. New vs Repeat Customers
# --------------------------------------
df["customer_type"] = np.where(df["repeat_purchase_rate"] > 0.5, "Repeat", "New")

new_repeat_summary = df.groupby("customer_type").agg(
    customer_count=("customer_id", "count"),
    avg_clv=("customer_lifetime_value", "mean"),
    avg_order_value=("avg_order_value", "mean"),
    avg_purchase_frequency=("purchase_frequency", "mean")
).reset_index()

print("=== New vs Repeat Customers ===")
print(new_repeat_summary)

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="customer_type", palette="Set2")
plt.title("New vs Repeat Customers")
plt.xlabel("Customer Type")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# Convert ALL columns that need numeric comparison
df["repeat_purchase_rate"] = pd.to_numeric(df["repeat_purchase_rate"], errors="coerce")
df["customer_lifetime_value"] = pd.to_numeric(df["customer_lifetime_value"], errors="coerce")
df["avg_order_value"] = pd.to_numeric(df["avg_order_value"], errors="coerce")
df["purchase_frequency"] = pd.to_numeric(df["purchase_frequency"], errors="coerce")
df["last_purchase_days"] = pd.to_numeric(df["last_purchase_days"], errors="coerce")  # ADD THIS
df["customer_satisfaction_score"] = pd.to_numeric(df["customer_satisfaction_score"], errors="coerce")
df["return_rate"] = pd.to_numeric(df["return_rate"], errors="coerce")

# --------------------------------------
# 2. Customer Segmentation
# --------------------------------------
conditions = [
    (df["customer_lifetime_value"] >= df["customer_lifetime_value"].quantile(0.75)) &
    (df["repeat_purchase_rate"] >= 0.75) &
    (df["purchase_frequency"] >= df["purchase_frequency"].median()),

    (df["customer_lifetime_value"] >= df["customer_lifetime_value"].quantile(0.50)) &
    (df["repeat_purchase_rate"] >= 0.60),

    (df["last_purchase_days"] > 30) &
    (df["repeat_purchase_rate"] < 0.50),

    (df["customer_satisfaction_score"] < 3.5) |
    (df["return_rate"] > df["return_rate"].quantile(0.75))
]

segment_names = [
    "Champions",
    "Loyal Customers",
    "At Risk",
    "Needs Attention"
]

df["derived_segment"] = np.select(conditions, segment_names, default="Regular")

segment_summary = df.groupby("derived_segment").agg(
    customer_count=("customer_id", "count"),
    avg_clv=("customer_lifetime_value", "mean"),
    avg_repeat_rate=("repeat_purchase_rate", "mean"),
    avg_purchase_frequency=("purchase_frequency", "mean")
).reset_index().sort_values(by="avg_clv", ascending=False)

print("\n=== Customer Segmentation ===")
print(segment_summary)

plt.figure(figsize=(10, 6))
sns.countplot(data=df, y="derived_segment", order=df["derived_segment"].value_counts().index, palette="viridis")
plt.title("Customer Segmentation")
plt.xlabel("Number of Customers")
plt.ylabel("Segment")
plt.tight_layout()
plt.show()
```

```python

# --------------------------------------
# 3. High-Value Customers
# --------------------------------------
# Customers in top 20% CLV

clv_threshold = df["customer_lifetime_value"].quantile(0.80)
high_value_customers = df[df["customer_lifetime_value"] >= clv_threshold].copy()

high_value_summary = high_value_customers[[
    "customer_id",
    "customer_lifetime_value",
    "avg_order_value",
    "purchase_frequency",
    "repeat_purchase_rate",
    "loyalty_tier",
    "segment",
    "geographic_region"
]].sort_values(by="customer_lifetime_value", ascending=False)

print("\n=== High-Value Customers ===")
print(high_value_summary.head(20))

plt.figure(figsize=(10, 6))
sns.histplot(df["customer_lifetime_value"], bins=30, kde=True, color="steelblue")
plt.axvline(clv_threshold, color="red", linestyle="--", label=f"Top 20% Threshold = {clv_threshold:.2f}")
plt.title("Customer Lifetime Value Distribution")
plt.xlabel("Customer Lifetime Value")
plt.ylabel("Count")
plt.legend()
plt.tight_layout()
plt.show()
```

```python


# --------------------------------------
# 4. Churn Probability Indicators
# --------------------------------------
# Create churn risk flags from available columns

df["risk_recency"] = np.where(df["last_purchase_days"] > 30, 1, 0)
df["risk_repeat"] = np.where(df["repeat_purchase_rate"] < 0.50, 1, 0)
df["risk_satisfaction"] = np.where(df["customer_satisfaction_score"] < 3.5, 1, 0)
df["risk_returns"] = np.where(df["return_rate"] > df["return_rate"].quantile(0.75), 1, 0)
df["risk_frequency"] = np.where(df["purchase_frequency"] < df["purchase_frequency"].median(), 1, 0)

df["churn_risk_score"] = (
    df["risk_recency"] +
    df["risk_repeat"] +
    df["risk_satisfaction"] +
    df["risk_returns"] +
    df["risk_frequency"]
)

df["churn_risk_level"] = pd.cut(
    df["churn_risk_score"],
    bins=[-1, 1, 3, 5],
    labels=["Low Risk", "Medium Risk", "High Risk"]
)

churn_summary = df.groupby("churn_risk_level").agg(
    customer_count=("customer_id", "count"),
    avg_last_purchase_days=("last_purchase_days", "mean"),
    avg_repeat_rate=("repeat_purchase_rate", "mean"),
    avg_satisfaction=("customer_satisfaction_score", "mean")
).reset_index()

print("\n=== Churn Probability Indicators ===")
print(churn_summary)

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="churn_risk_level", order=["Low Risk", "Medium Risk", "High Risk"], palette="coolwarm")
plt.title("Churn Risk Levels")
plt.xlabel("Risk Level")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()


# --------------------------------------
# Optional: export outputs
# --------------------------------------
new_repeat_summary.to_csv("new_vs_repeat_summary.csv", index=False)
segment_summary.to_csv("customer_segmentation_summary.csv", index=False)
high_value_summary.to_csv("high_value_customers.csv", index=False)
churn_summary.to_csv("churn_summary.csv", index=False)
```
